# Exploratory Data Analysis (EDA)
## Student Performance Analysis

**Objective:** To explore student data, examine
data quality, calculate summary statistics,
and visualize patterns in study hours, attendance,
and marks.

## Step 1: Dataset Creation

A sample dataset of eight students was created
with four columns: Student, Study_Hours,
Attendance, and Marks.

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Study_Hours": [2, 4, 1, 5, 3, 6, 2, 4],
    "Attendance": [70, 85, 60, 95, 75, 98, 65, 88],
    "Marks": [45, 70, 35, 90, 55, 95, 40, 78]
}

df = pd.DataFrame(data)

print("Dataset created successfully!")
display(df)
```

## Step 2: Understanding the Dataset

The dataset shape, column names, data types, and descriptive statistics were examined using Pandas.

```python
print("Dataset shape:", df.shape)

print("\nColumn names:")
print(df.columns)

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
display(df.describe())
```

## Step 3: Data Quality Checks

Missing values and exact duplicate rows were checked to assess data completeness.

```python
print("Missing values in each column:")
print(df.isnull().sum())

print("\nTotal duplicate rows:")
print(df.duplicated().sum())
```

## Step 4: Correlation Analysis

Pearson correlation was calculated to examine the linear relationship between study hours and marks.

```python
print("Study Hours and Marks:")
display(df[["Study_Hours", "Marks"]])

print("\nCorrelation between study hours and marks:")
print(df["Study_Hours"].corr(df["Marks"]))
```


##Step 5: Scatter Plot — Study Hours vs Marks
The scatter plot visualizes the relationship between study hours and marks.

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Study_Hours"], df["Marks"])

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.grid(True)
plt.show()
```



## Step 6: Histogram — Distribution of Marks

The histogram displays the distribution of marks across intervals.



```python
plt.figure(figsize=(8, 5))

plt.hist(df["Marks"], bins=5, edgecolor="black")

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```



## Step 7: Bar Chart — Student Attendance

The bar chart compares attendance percentages across students.


```python
plt.figure(figsize=(8, 5))

plt.bar(df["Student"], df["Attendance"])

plt.title("Student Attendance")
plt.xlabel("Student")
plt.ylabel("Attendance (%)")

plt.show()
```


## Step 8: EDA Summary

The main statistics and data quality results were summarized.

## Conclusion

The sample contains eight student records. No missing values or exact duplicate rows were detected. A strong positive correlation was observed between study hours and marks in this sample. The results are illustrative and cannot be generalized to all students.




```python
print("EDA SUMMARY")
print("-" * 30)

print("Total students:", len(df))
print("Average study hours:", round(df["Study_Hours"].mean(), 2))
print("Average attendance:", round(df["Attendance"].mean(), 2), "%")
print("Average marks:", round(df["Marks"].mean(), 2))

print("Highest marks:", df["Marks"].max())
print("Lowest marks:", df["Marks"].min())

print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())

correlation = df["Study_Hours"].corr(df["Marks"])
print("Study hours-mark correlation:", round(correlation, 3))
```
